## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

## Observed
- Traffic decay is widespread across the dataset, with 50\% of pages experiencing a traffic loss of 33.5% or worse trend_pct median is -33.5\%, mean is -4.79%.

- A substantial portion of content sits in striking distance of top rankings, with 50% of pages holding an average rank of position 10.8 or better (avg_position median is 10.8).

- The raw metrics contain extreme upper-bound outliers and data noise, including maximum click-through rates and engagement rates recorded at 100% and scroll rates reaching up to 300%.

## Directional
- Relative scoring rank orders pages by their expected potential return on editing effort, indicating which decaying pages show higher opportunity relative to others.

- Trend indicators (trend_pct median of -33.5%) identify trajectories of decaying traffic versus stable or growing content across individual page units where 1 text row = 1 text page.

- High-opportunity signals (such as striking-distance average positions with a median rank of 10.8) point toward pages that are closer to page-one visibility, prioritizing them over low-impact or deep-page content.

## Decision Support

# What it will say:

- PRIORITY RANKING: Ranks pages by content refresh opportunity based on historical performance signals (such as striking-distance position ranks and trend decay rates).

RESOURCE ALLOCATION: Identifies which top subset of decaying content offers the highest potential return on editing effort.

TRIAGE & FILTERING: Flags high-volume pages experiencing traffic decay so editors can prioritize re-optimizing them over low-impact content.

# What it will NEVER say:

CAUSAL PROOF: Does not guarantee that updating a page will directly cause Google rankings or traffic to increase.

PREDICTING GOOGLE: Does not model or predict Google's proprietary search engine ranking algorithm.

AUTOMATED ACTION: Does not automatically rewrite content or declare pages broken without human editorial review and approval.
