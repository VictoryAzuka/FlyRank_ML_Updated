```python
import os
import requests

base_github_url = 'https://raw.githubusercontent.com/VictoryAzuka/FlyRank_ML_Updated/main/'

# --- Download skills/README.md ---
skill_readme_path = 'skills/README.md'
skill_readme_url = base_github_url + skill_readme_path

# Ensure the directory exists
os.makedirs(os.path.dirname(skill_readme_path), exist_ok=True)

print(f"Downloading {skill_readme_path} from {skill_readme_url}...")
response_skills_readme = requests.get(skill_readme_url)
response_skills_readme.raise_for_status() # Raise an exception for bad status codes

# Save the file locally
with open(skill_readme_path, 'wb') as f:
    f.write(response_skills_readme.content)
print(f"Successfully downloaded and saved to {skill_readme_path}")

print("\n" + "="*50 + "\n") # Separator for clarity

# --- Download framing-ml-problems/SKILL.md ---
framing_skill_path = 'skills/framing-ml-problems/SKILL.md' # Corrected path
framing_skill_url = base_github_url + framing_skill_path

# Ensure the directory exists
os.makedirs(os.path.dirname(framing_skill_path), exist_ok=True)

print(f"Downloading {framing_skill_path} from {framing_skill_url}...")
response_framing_skill = requests.get(framing_skill_url)
response_framing_skill.raise_for_status() # Raise an exception for bad status codes

# Save the file locally
with open(framing_skill_path, 'wb') as f:
    f.write(response_framing_skill.content)
print(f"Successfully downloaded and saved to {framing_skill_path}")
```
