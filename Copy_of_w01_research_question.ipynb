```python
import os
import requests

# Define the directory and file path
data_dir = 'data/raw'
csv_filename = 'content_refresh_anonymized.csv'
csv_filepath = os.path.join(data_dir, csv_filename)

# Create the directory if it doesn't exist
os.makedirs(data_dir, exist_ok=True)

# Define the URL for the raw CSV file on GitHub
github_csv_url = 'https://raw.githubusercontent.com/VictoryAzuka/FlyRank_ML_Updated/main/data/raw/content_refresh_anonymized.csv'

print(f"Downloading {csv_filename} from {github_csv_url}...")
response = requests.get(github_csv_url)
response.raise_for_status() # Raise an exception for bad status codes

# Save the file locally
with open(csv_filepath, 'wb') as f:
    f.write(response.content)

print(f"Successfully downloaded and saved to {csv_filepath}")
```
