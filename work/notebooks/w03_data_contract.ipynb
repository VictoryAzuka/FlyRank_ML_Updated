## 1. Unit of analysis + time window

### Unit of Analysis
- **Unit**: One row represents a single pseudonymized content item (`content_id`) for a specific `client_id`.
- **Verify Grain**: We must confirm that `content_id` uniquely identifies each row.


### Time Window
- The metrics in this dataset capture a trailing 90-day aggregate window up to the extraction point.
- We will check the distribution of any timestamp or temporal indicators.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# 1. Verify Grain (Unit of Analysis)
total_rows = len(df)
unique_content_ids = df['content_id'].nunique()
print(f"Total rows: {total_rows}")
print(f"Unique content IDs: {unique_content_ids}")

# Check if content_id is a unique key
is_unique = total_rows == unique_content_ids
print(f"Is 'content_id' uniquely identifying each row? {is_unique}")

# Grouping check
grain_check = df.groupby('content_id').size().reset_index(name='count')
duplicates = grain_check[grain_check['count'] > 1]
print(f"Number of duplicate content_id rows: {len(duplicates)}")

# 2. Check the Shape and Columns
print(f"Dataset Shape: {df.shape}")
display(df.head())
```

## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

```python
# Let's inspect the categorical columns (object/string type) in the dataset
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
print("Categorical Columns identified:", cat_cols)

```

### Fields Classification

Based on the analysis of categorical columns and the `skills/flyrank/flyrank-data/SKILL.md` rules, we classify them as follows:

1. **Context**
   - `content_id`: Unique identifier for each item. Used for grouping/joining, never as a feature.
   - `client_id`: Unique identifier for each client. Used for split strategies (grouped split), never as a feature.

2. **Label / Proxy**
   - `is_declining_label`: The primary target flag we aim to predict.
   - `trend_direction`: The direct basis for computing `is_declining_label`. Using this as a feature causes a massive data leak (label trap).

3. **Feature**
   - `content_type`: Strong feature indicating the layout or purpose of the content. Note: missingness patterns in other features may depend on this field.

4. **Excluded**
   - `trend_direction`: (Also classified under Label/Proxy) Excluded from features because it is derived from `trend_pct` and directly leaks the target label `is_declining_label`.

## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
import pandas as pd
import numpy as np

print("=== 1. GRAIN CHECK ===")
# Grain: Is content_id unique?
grain_violations = df.groupby('content_id').size().reset_index(name='count').query('count > 1')
print(f"Rows violating unique grain (content_id): {len(grain_violations)}")

print("\n=== 2. COUNTS CHECK ===")
total_rows = len(df)
unique_clients = df['client_id'].nunique()
print(f"Total rows in dataset: {total_rows}")
print(f"Total unique clients: {unique_clients}")

print("\n=== 3. MISSINGNESS CHECK (Percentage of Nulls per Column) ===")
missingness = df.isnull().mean()
print(missingness[missingness > 0].sort_values(ascending=False))

print("\n=== 4. WINDOWS / RANGES CHECK ===")
# Let's verify some numerical boundaries and categories
print(f"Trend PCT range: {df['trend_pct'].min()}% to {df['trend_pct'].max()}%")
```

```python
# Let's check some statistics of trend_pct to see the distribution of these extreme percentages
print(df['trend_pct'].describe(percentiles=[0.5, 0.9, 0.95, 0.99, 0.999]))
```
